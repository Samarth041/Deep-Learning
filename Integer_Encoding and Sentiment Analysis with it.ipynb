### Sentiment Analysis through Integer Encoding Method

<h4>How we do integer encoding</h4>

```python
import numpy as np

#input data contains 10 sentences
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'modi ji ki jai',
		'inquilab zindabad']
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer

tokenizer = Tokenizer(oov_token='<nothing>') 
#oov_token stands for Out-Of-Vocabulary token.
# It is a special token used by the tokenizer to represent words that were not seen
#during training (i.e., words that are not in the tokenizer's vocabulary).
```

```python
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
```

```python
tokenizer.word_counts #kon sa word kitni baar aya
```

```python
tokenizer.document_count #how many sentences
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences #applying padding
```

```python
sequences = pad_sequences(sequences,padding='post')
```

```python
sequences
```

<h4>Applying on Dataset imbd</h4>

```python
from keras.datasets import imdb
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
X_train #in vocabulary we have assigned number to each words thats why we are seeing numbers
```

```python
X_train[0] #this data is already integer encoded and already preprocessed
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

maxlen = 200

X_train = pad_sequences(X_train, maxlen=maxlen)
X_test = pad_sequences(X_test, maxlen=maxlen)

print(X_train.dtype)
print(X_train.shape)
```

```python
X_train[0]
```

```python
model = Sequential()

model.add(SimpleRNN(32,input_shape=(200,1),return_sequences=False)) #(200->timestep,1 feature at a time)
model.add(Dense(1,activation='sigmoid'))

model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=5,validation_data=(X_test,y_test))
```

<h4>Testing the performance of model by predicting something with it</h4>

```python

word_index = imdb.get_word_index()

# Create a reverse word index to map words back to integers
# Remember that the IMDB dataset reserves indices 0, 1, and 2 for padding, start of sentence, and unknown words.
# So we need to add 3 to each word index.
word_to_id = {k: v + 3 for k, v in word_index.items()}
id_to_word = {value: key for key, value in word_to_id.items()}

# Define a function to preprocess a new review
def preprocess_review(review_text, word_to_id, maxlen=50):
    # Convert review to lowercase and split into words
    words = review_text.lower().split()

    # Convert words to their corresponding IMDB indices
    # Handle words not in the vocabulary by mapping them to index 2 (unknown)
    indexed_review = [word_to_id.get(word, 2) for word in words]

    # Pad the sequence to the fixed length (maxlen=50)
    padded_review = pad_sequences([indexed_review], padding='post', maxlen=maxlen)

    # Reshape for the SimpleRNN input layer (batch_size, timesteps, features)
    # The model was built with input_shape=(50,1), so we add a feature dimension
    reshaped_review = np.expand_dims(padded_review, axis=-1)

    return reshaped_review


# Example new negative review
negative_review = "This film was terrible, a complete waste of time. I would not recommend it at all."


processed_negative_review = preprocess_review(negative_review, word_to_id, maxlen=50)

# Predict the sentiment for the negative review
negative_prediction = model.predict(processed_negative_review)

# Interpret the prediction
if negative_prediction[0][0] > 0.5:
    negative_sentiment = "Positive"
else:
    negative_sentiment = "Negative"

print(f"Review: '{negative_review}'")
print(f"Predicted sentiment score: {negative_prediction[0][0]:.4f}")
print(f"Predicted sentiment: {negative_sentiment}")
```

## now sentiment analysis with embedding

```python
from tensorflow.keras.datasets import imdb

(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
X_train.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, Dense

model = Sequential()

model.add(Embedding(
    input_dim=10000,
    output_dim=2,
    input_length=50   # or just omit it
))

model.add(SimpleRNN(32, return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```

### Prediction

```python

word_index = imdb.get_word_index()


word_to_id = {k: v + 3 for k, v in word_index.items()}
id_to_word = {value: key for key, value in word_to_id.items()}

# Define a function to preprocess a new review
def preprocess_review(review_text, word_to_id, maxlen=50):
    # Convert review to lowercase and split into words
    words = review_text.lower().split()

    # Convert words to their corresponding IMDB indices
    # Handle words not in the vocabulary by mapping them to index 2 (unknown)
    indexed_review = [word_to_id.get(word, 2) for word in words]

    # Pad the sequence to the fixed length (maxlen=50)
    padded_review = pad_sequences([indexed_review], padding='post', maxlen=maxlen)

    # Reshape for the SimpleRNN input layer (batch_size, timesteps, features)
    # The model was built with input_shape=(50,1), so we add a feature dimension
    reshaped_review = np.expand_dims(padded_review, axis=-1)

    return reshaped_review

# Example new positive review
positive_review = "This movie was an absolute masterpiece! I loved every single moment, truly captivating."

# Example new negative review
negative_review = "This film was terrible, a complete waste of time. I would not recommend it at all."

# Preprocess the positive review
processed_positive_review = preprocess_review(positive_review, word_to_id, maxlen=50)

# Predict the sentiment for the positive review
positive_prediction = model.predict(processed_positive_review)

# Interpret the prediction (sigmoid output: > 0.5 for positive)
if positive_prediction[0][0] > 0.5:
    positive_sentiment = "Positive"
else:
    positive_sentiment = "Negative"

print(f"Review: '{positive_review}'")
print(f"Predicted sentiment score: {positive_prediction[0][0]:.4f}")
print(f"Predicted sentiment: {positive_sentiment}\n")

# Preprocess the negative review
processed_negative_review = preprocess_review(negative_review, word_to_id, maxlen=50)

# Predict the sentiment for the negative review
negative_prediction = model.predict(processed_negative_review)

# Interpret the prediction
if negative_prediction[0][0] > 0.5:
    negative_sentiment = "Positive"
else:
    negative_sentiment = "Negative"

print(f"Review: '{negative_review}'")
print(f"Predicted sentiment score: {negative_prediction[0][0]:.4f}")
print(f"Predicted sentiment: {negative_sentiment}")
```
